```python
import pandas as pd
import matplotlib.pyplot as plt

#  Load the dataset
# This file is usually cleaner, but I use 'on_bad_lines' just in case
df = pd.read_csv('../data/raw/phishing_email.csv')

#  Check the balance 
print(f"Total Emails: {len(df)}")
print("\nClass Balance:")
print(df['label'].value_counts())

# Visualize
plt.figure(figsize=(6,4))
df['label'].value_counts().plot(kind='bar', color=['green', 'red'])
plt.title('Email Dataset Balance (0=Safe, 1=Phishing)')
plt.xticks(rotation=0)
plt.show()

# Preview the data
print("\n--- Sample Email Text ---")
print(df['text_combined'].iloc[0][:300])  
```
